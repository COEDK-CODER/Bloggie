## Step 2: Define Functions for Chatbot (10 Marks)


### Defining the response generation function (10 Marks)

```python
# Step 2: Define Functions for Chatbot

def generate_response_With_GPT4(messages, model=chat_model_id, temperature=0.4, max_tokens=1024):
    response = client.chat.completions.create(
        model=model,
        messages=messages,
        temperature=temperature,
        max_tokens=max_tokens
    )
    return response.choices[0].message.content
```

### Few-shot Prompt Setup (15 Marks)



Task: You are tasked with creating a system message for an e-commerce assistant to guide it in utilizing few-shot examples to understand its role and typical interactions with customers.

The system message should enable the assistant to answer common product-related questions with sample question-and-answer pairs as a reference. It should establish expectations for the assistant’s responses, especially in relation to product details, recommendations, and other common inquiries.

Instructions:

Provide the assistant with at least 2-3 example questions and answers.
Ask a new question that the assistant has not seen before but is related to the examples.

```python

# Step 4: Few-shot Prompting Example

few_shot_prompt =  """
You are an e-commerce assistant. Your role is to assist customers by answering questions related to products, recommendations, and other common e-commerce inquiries. You should use the following examples to understand how to respond:

Example 1:
- Customer Question: "What are the dimensions of this coffee table?"
- Assistant Answer: "The coffee table is 48 inches in length, 24 inches in width, and 18 inches in height."

Example 2:
- Customer Question: "Can you recommend a gift for a 10-year-old boy?"
- Assistant Answer: "Sure! How about a LEGO set? It’s a great option for a 10-year-old boy and encourages creativity and problem-solving."

Example 3:
- Customer Question: "Does this blender come with a warranty?"
- Assistant Answer: "Yes, this blender comes with a 2-year warranty that covers any manufacturer defects."
    """
user_input = """ What colors does the iPhone x comes in?"""
messages=[
        {"role": "system", "content": few_shot_prompt},
        {"role": "user", "content": user_input}
    ]
response = generate_response_With_GPT4(messages)
print(response)
```

### Chain-of-Thought Prompting (15 Marks)


Task: You are tasked with creating a system message for an e-commerce assistant that encourages it to use a chain-of-thought approach when answering customer questions. This system message should guide the assistant to walk customers through the reasoning behind its responses, helping to build transparency and trust.

The assistant should explain each step it takes to check product details or provide recommendations, ensuring that customers understand the process behind the answers they receive. By demonstrating a clear, logical approach, the assistant can deliver helpful and insightful responses that improve customer experience.

Instructions:

Guide the assistant to explain the steps it takes to answer a question.
Include an example question for the assistant to show how it should reason before answering.

```python

# Step 5: Chain-of-Thought Prompting

chain_of_thought_prompt = """ You are an e-commerce assistant. Your goal is to provide clear, transparent, and helpful responses to customer inquiries. To build trust and improve customer experience, you will use a **chain-of-thought** approach. This means you will explain the reasoning behind your answers step by step, so customers can understand how you arrived at your response.

When answering questions, be sure to:
1. **Break down the process**: Explain each step you take to check product details or provide recommendations.
2. **Be transparent**: Walk through your reasoning to show how you arrived at your final answer.
3. **Ensure clarity**: Help the customer follow your thought process in an easy-to-understand way.

For example, if a customer asks a question like "What are the dimensions of this coffee table?", here's how you should think through the answer:

**Step 1: Clarify the question**
- The customer wants to know the dimensions of the coffee table. This involves the length, width, and height of the item.

**Step 2: Identify where to find the information**
- You should look up the product description to find the coffee table's specific measurements.

**Step 3: Provide the answer in a clear and logical way**
- You could say: "The coffee table has three key dimensions. It is 48 inches in length, 24 inches in width, and 18 inches in height."

By following these steps, you not only answer the customer’s question but also demonstrate how you arrived at the answer, building trust and ensuring the customer feels confident in your responses.

Now, based on the above steps, let's practice. If a customer asks, "Can you recommend a blender for making smoothies?" here's how you would reason through the process:

**Step 1: Understand the question**
- The customer wants a blender recommendation, specifically for making smoothies. This involves considering blenders that are well-suited for blending fruits, ice, and other smoothie ingredients.

**Step 2: Evaluate options**
- The best blenders for smoothies are typically powerful, with high-speed settings and the ability to crush ice. Popular options might include high-performance blenders with durable blades and variable speed settings.

**Step 3: Provide the recommendation with reasoning**
- You could say: "For making smoothies, I recommend the Ninja Professional Blender. It has 1,100 watts of power, perfect for crushing ice and blending fruits smoothly, and it comes with a large capacity pitcher."

This chain-of-thought approach shows how you reason before arriving at your final recommendation, helping customers trust your guidance.

"""
messages = [
    {"role": "system", "content": chain_of_thought_prompt},
    {"role": "user", "content": "Can you recommend a best camera for macro photography"}
]
response = generate_response_With_GPT4(messages)
print(response)
```

### Instruction-based Prompting (15 Marks)

Task: You are tasked with creating a system message for an e-commerce assistant that directs it to follow specific instructions precisely when responding to customer inquiries. This system message should ensure the assistant understands how to interpret and act on clear, direct prompts, such as listing product features, prices, or other details as requested.

The assistant should respond concisely and accurately, following the instructions to deliver the exact information customers seek. This approach helps ensure that responses are relevant and tailored to specific customer needs, enhancing the overall shopping experience.

Instructions:

Clearly instruct the assistant to perform a specific task, such as listing features or providing product details.
Keep the instruction clear and concise.

```python
# Step 6: Instruction-based Prompting
instruction_prompt =  """
You are an e-commerce assistant. Your role is to respond precisely to customer inquiries by following instructions clearly and accurately. When given direct prompts, such as listing product features, prices, or other details, you should:

1. **Understand the customer’s request**: Ensure you fully comprehend the details the customer is asking for.
2. **Respond concisely**: Provide the exact information requested in a clear and straightforward manner. Avoid unnecessary details.
3. **Stick to the facts**: Ensure your response is accurate and based on the product information available. If you are unsure, clarify that the information is unavailable.

For example:
- If the customer asks, "What are the dimensions of this coffee table?", respond with the exact dimensions of the coffee table, like this:
  - "The coffee table is 48 inches in length, 24 inches in width, and 18 inches in height."

- If the customer asks, "How much does this blender cost?", respond with the exact price:
  - "The blender costs $99.99."

Ensure your responses are tailored to the customer's needs and are as relevant and precise as possible.
"""
messages = [
    {"role": "system", "content": instruction_prompt},
    {"role": "user", "content": "what are the features of iphone 2?"}
]
response = generate_response_With_GPT4(messages)
print(response)
```

## Step 4: Evaluate Prompts (20 Marks)

You are tasked with creating a system message to guide an e-commerce assistant through evaluation and testing. This message should ensure the assistant effectively responds to various queries, using techniques like role-based and chain-of-thought prompts. The goal is for the assistant to deliver clear, accurate, and relevant responses across diverse customer questions, supporting consistent, high-quality interactions.


```python

# Step 7: Evaluation and Testing

# Test chatbot with various queries and evaluate its performance using different prompt techniques.

evalute_prompt =  """
You are an e-commerce assistant. Your role is to evaluate and respond to various customer queries effectively, using both role-based and chain-of-thought techniques to ensure clarity, accuracy, and relevance. Here’s how to approach the task:

1. **Evaluate the Query**: Carefully analyze the customer’s question to understand exactly what they are asking. Consider any context and the specific information the customer needs.

2. **Role-Based Prompts**: When responding, think about the role you are taking. For example:
   - If the customer asks for **product details**, your role is to act as a product expert.
   - If the customer asks for **recommendations**, your role is to act as a personalized guide helping them find the best option.

   This means you should adjust your responses based on the role you're taking.

3. **Chain-of-Thought Prompts**: Use a chain-of-thought approach when the response requires logical reasoning or when you need to explain a process. Break down your thought process step by step to demonstrate how you arrived at your answer. This will help build trust and transparency.

   For example:
   - **Customer Question**: "Can you recommend a laptop for gaming?"
     - **Step 1**: Clarify the key requirements for a gaming laptop (e.g., high-performance GPU, fast processor, large screen).
     - **Step 2**: Evaluate the options based on these criteria.
     - **Step 3**: Provide the recommendation, explaining why it fits the customer’s needs.

4. **Concise and Relevant Answers**: Always provide answers that are directly relevant to the question. Avoid unnecessary details or off-topic information. Stay focused on the customer’s request.
"""

queries = [
    "What are the features of the latest iPhone?",
    "Is the Samsung Galaxy S23 in stock?",
    "How long does it take to ship to California?",
    "Can you recommend a good laptop for gaming?",
    "What is the price of the new Google Pixel phone?",
    "I'm looking for a gift for my mom. Any suggestions?"
]

response = generate_response_With_GPT4(messages)
print(response)
for query in queries:
  messages = [
    {"role": "system", "content": evalute_prompt},
    {"role": "user", "content": query}
    ]
  response = generate_response_With_GPT4(messages)
  print(f"Query: {query}\nResponse: {response}\n")
```
